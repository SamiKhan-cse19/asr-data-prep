```python
import os
import torchaudio
from yt_download import download_urls
from silero_vad import load_silero_vad, read_audio, get_speech_timestamps
```

```python
audio_files = [
    "data/sWhIX5A-9pY.mp3"
]

model = load_silero_vad()

wav = read_audio(audio_files[0])

speech_timestamps = get_speech_timestamps(
    wav,
    model,
    threshold=0.5,
    speech_pad_ms=50,
    min_speech_duration_ms=3000,
    max_speech_duration_s=30,
    min_silence_duration_ms=600,
    # return_seconds=True,  # Return speech timestamps in seconds (default is samples)
)
```

```python
sample_rate = 16000  

if wav.dim() == 1:
    wav = wav.unsqueeze(0)
```

```python
timestamp = speech_timestamps[0]
start = timestamp["start"]
end = timestamp["end"]
    
segment = wav[:, start:end] 
        
output_path = os.path.join("chunks", f"chunk_test.mp3")
torchaudio.save(output_path, segment, sample_rate, format="mp3")
```
